## 29장. 실수 점검과 제한 시간 운영

형태·실행 순서·오류 대응·저장·최종 검토를 연습한다.

```python
import numpy as np
from lab import split_task, preprocessor

X_train, X_val, X_test, y_train, y_val, y_test = split_task()
assert X_train.index.equals(y_train.index)
assert set(X_train.index).isdisjoint(X_val.index)
assert set(X_train.index).isdisjoint(X_test.index)
assert "churn" not in X_train.columns
prep = preprocessor()
train = prep.fit_transform(X_train)
val = prep.transform(X_val)
assert train.shape[1] == val.shape[1]
assert np.isfinite(train).all() and np.isfinite(val).all()
assert len(y_train) == len(train)
print("Input, split and transformation checks passed")
```
