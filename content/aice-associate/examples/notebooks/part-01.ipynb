## 03장. 실습에 필요한 Python과 NumPy

자료형·인덱싱·함수·배열의 형태를 익힌다.

```python
import numpy as np

fees = np.array([20.0, 30.0, 40.0])
X = fees.reshape(-1, 1)
y = np.array([0, 1, 0])
assert X.shape == (3, 1)
assert y.shape == (3,)
mask = fees >= 30
assert fees[mask].tolist() == [30.0, 40.0]
assert np.isclose(fees.mean(), 30.0)
print(X.shape, y.shape, fees[mask])
```
